```python
def factoria (n):
    if n == 0 or n ==1:
        return 1
    if n < 0:
        raise ValueError("Factorial is undefined ")
    return n*factoria(n-1)
```

```python
def gcd(num1: int, num2: int) -> int:
    limit = min(num1, num2)
    gcd = 1

    for divisor in range(1, limit + 1):
        if num1 % divisor == 0 and num2 % divisor == 0:
            gcd = divisor

    return gcd


def lcm(num1: int, num2: int) -> int:
    return (num1 * num2) // gcd(num1, num2)


print(lcm(12, 18))
```

```python
def count_words(words: list[str]) -> dict[str, int]:
    """Count the occurrences of each unique word.

    Args:
        words: A list of words.

    Returns:
        A dictionary containing each word and its frequency.
    """
    unique_words = set(words)
    counts = {word: 0 for word in unique_words}

    for i in unique_words:
        for j in words:
            if i == j:
                counts[i] += 1

    return counts


words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]

print(count_words(words))
```
