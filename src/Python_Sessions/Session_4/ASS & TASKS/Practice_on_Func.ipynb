2 - Twin Primes Under 1000

```python

def is_Prime(num : int):
    if num <2 :
        return False
    for i in range (2,int(num**0.5)+1):
        if num % i == 0:
            return False
        

    return True


def twin_prime():
    
    print(f"|(*_*){'TWIN PRIME':^30}(^_^)|\n \n")
    for i in range(1,1000):
        if is_Prime(i) and is_Prime(i+2):
            print(f" {i:>3} {'and':^5} {i+2:<3}")

twin_prime()
```

3- Factors Of Numbers

```python
def Factors_for_primes(num : int = 20):
    result = []
    for i in range(2,num):
        while num % i ==0 and is_Prime(i):            
            result.append(i)
            num //= i

    print(result)
num = int(input("BROOO Enter num to get it's Factors : (^_^)"))
Factors_for_primes(num)
```

4- Decimal To Binary

5 - Perfect Nnumber : 

```python
def is_Perfect(num:int) :
    devisor =[]
    total = 0
    for i in range(1 , num):
        if num % i == 0 :
            devisor.append(i)
            total += i

    # print(devisor)
    if total == num:
        return True
    else:
        return False

def print_Perfect_nums(start : int, end:int):
    for i in range(start , end):
        if is_Perfect(i):
            print(f"{i}")

print_Perfect_nums(0,100)
```
